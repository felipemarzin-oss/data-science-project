#### Supprimer les personnes identiques 

```python
sans_doublons = df.drop_duplicates().copy()
print(sans_doublons.duplicated().sum())
print(sans_doublons.shape)
```

## Réponses styles musicaux : cochés partout ou 0 réponses
#### Personnes qui ont baclés leurs réponses

```python
# transformation des réponses en 0 et 1 
dataset_code = sans_doublons.copy()
style = [f"Q5_{i}" for i in range(1, 14)]

codage = {
    '0': 0,
    '1': 1,
    'Variété Française': 1,
    'Pop / Rock': 1,
    'Rap / Hip-Hop': 1,
    'Classique': 1,
    'Jazz': 1,
    'Dance': 1,
    'Electronique': 1,
    'Métal': 1,
    'R&B': 1,
    'Soul': 1,
    'Reggae': 1,
    'Musiques du monde': 1,
    'Autres genres': 1
}

for colonne in style:
    reponses = dataset_code[colonne].astype("string").str.strip()
    dataset_code[colonne] = reponses.map(codage)

dataset_code[style].head()
```

#### Personnes qui ont cochés tous les styles de musique
Une personne. Cette personne écoute de la musique plusieurs fois par jours ==> voir si suppression de cette ligne 
(Q3 = De manière générale, à quelle fréquence consommez-vous de la musique ou des vidéo clips , quel que soit le support ou le moyen de consommation ?)

```python
dataset_code[style][dataset_code[style].sum(axis=1) > 12]
```

#### Personnes qui n'ont rien coché

```python
dataset_code[style][(dataset_code[style].sum(axis=1) == 0) & (dataset_code[style].notna().all(axis=1))]
```
